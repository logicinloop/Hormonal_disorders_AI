```python
!pip install -q xgboost shap
from google.colab import files
import zipfile, glob, os

up = files.upload()

# unzip every zip you uploaded
for f in list(up):
    if f.lower().endswith(".zip"):
        folder = "data_" + os.path.splitext(f)[0].replace(" ", "_").replace("+", "_")
        with zipfile.ZipFile(f) as z:
            z.extractall(folder)

def find(name):
    hits = [p for p in glob.glob("**/*.csv", recursive=True) if os.path.basename(p).lower() == name.lower()]
    return hits[0] if hits else None

own   = find("balanced_multidisease_model_dataset.csv")
brfss = find("diabetes_binary_5050split_health_indicators_BRFSS2015.csv")
uci   = find("diabetes_data_upload.csv")

print("own  :", own)
print("brfss:", brfss)
print("uci  :", uci)
print("\nAll CSVs found:")
for p in glob.glob("**/*.csv", recursive=True):
    print("  ", p)
```
